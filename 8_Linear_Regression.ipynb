### Linear Regression Implementation

This notebook demonstrates a simple implementation of the Linear Regression (LR) algorithm from scratch using NumPy, along with a visualization of the fitted line using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the Linear Regression class
class LinearRegression:
    def __init__(self, learning_rate=0.01, epochs=1000):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        num_samples, num_features = X.shape
        # Initialize parameters
        self.weights = np.zeros(num_features)
        self.bias = 0

        # Gradient Descent
        for _ in range(self.epochs):
            y_predicted = np.dot(X, self.weights) + self.bias

            # Calculate gradients
            dw = (1 / num_samples) * np.dot(X.T, (y_predicted - y))
            db = (1 / num_samples) * np.sum(y_predicted - y)

            # Update parameters
            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict(self, X):
        return np.dot(X, self.weights) + self.bias

# 2. Generate some sample data
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X.squeeze() + np.random.randn(100)

# 3. Train the model
model = LinearRegression(learning_rate=0.1, epochs=500)
model.fit(X, y)
predictions = model.predict(X)

# 4. Plot the results
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(X, y, color='blue', label='Data Points')
ax.plot(X, predictions, color='red', linewidth=2, label=f'LR Line (y = {model.weights[0]:.2f}x + {model.bias:.2f})')
ax.set_title('Linear Regression from Scratch')
ax.set_xlabel('X')
ax.set_ylabel('y')
ax.legend()
plt.show()
```
